```python
# Ejercicio 2

class Empresa:
  """Clase que representa una empresa."""

  empresa = "UCM"  # Atributo de clase

class Empleado:
  """Clase que representa a un empleado de la empresa."""

  def __init__(self, nombre, rut, fecha_nacimiento, unidad_administrativa):
    """Inicializa un nuevo empleado.

    Args:
      nombre: El nombre del empleado.
      rut: El RUT del empleado.
      fecha_nacimiento: La fecha de nacimiento del empleado en formato yyyy-mm-dd.
      unidad_administrativa: La unidad administrativa a la que pertenece el empleado.
    """
    self.nombre = nombre
    self.rut = rut
    self.fecha_nacimiento = fecha_nacimiento
    self.unidad_administrativa = unidad_administrativa

  @property
  def nombre(self):
    """Devuelve el nombre del empleado."""
    return self._nombre

  @nombre.setter
  def nombre(self, nombre):
    """Establece el nombre del empleado."""
    self._nombre = nombre

  @property
  def rut(self):
    """Devuelve el RUT del empleado."""
    return self._rut

  @rut.setter
  def rut(self, rut):
    """Establece el RUT del empleado."""
    self._rut = rut

  @property
  def fecha_nacimiento(self):
    """Devuelve la fecha de nacimiento del empleado."""
    return self._fecha_nacimiento

  @fecha_nacimiento.setter
  def fecha_nacimiento(self, fecha_nacimiento):
    """Establece la fecha de nacimiento del empleado."""
    self._fecha_nacimiento = fecha_nacimiento

  @property
  def unidad_administrativa(self):
    """Devuelve la unidad administrativa del empleado."""
    return self._unidad_administrativa

  @unidad_administrativa.setter
  def unidad_administrativa(self, unidad_administrativa):
    """Establece la unidad administrativa del empleado."""
    self._unidad_administrativa = unidad_administrativa

  def calcular_edad(self):
    """Calcula la edad del empleado a partir de su fecha de nacimiento.

    Returns:
      La edad del empleado en años.
    """
    from datetime import datetime, date

    fecha_nacimiento = datetime.strptime(self.fecha_nacimiento, "%Y-%m-%d")
    fecha_actual = date.today()
    edad = fecha_actual.year - fecha_nacimiento.year
    if fecha_actual.month < fecha_nacimiento.month or \
       (fecha_actual.month == fecha_nacimiento.month and
        fecha_actual.day < fecha_nacimiento.day):
      edad -= 1
    return edad

  @classmethod
  def from_dict(cls, datos_empleado):
    """Construye una instancia de Empleado a partir de un diccionario.

    Args:
      datos_empleado: Un diccionario que contiene los datos del empleado.

    Returns:
      Una instancia de Empleado.
    """
    return cls(**datos_empleado)

  def __str__(self):
    """Devuelve una representación en cadena del empleado."""
    return f"Nombre: {self.nombre}, RUT: {self.rut}, Fecha de nacimiento: {self.fecha_nacimiento}, Unidad administrativa: {self.unidad_administrativa}"

  def __repr__(self):
    """Devuelve una representación completa del empleado."""
    return f"Empleado(nombre='{self.nombre}', rut='{self.rut}', fecha_nacimiento='{self.fecha_nacimiento}', unidad_administrativa='{self.unidad_administrativa}')"

# Ejemplo de uso
empleado1 = Empleado(nombre="Juan Pérez", rut="12345678-9", fecha_nacimiento="1980-01-01", unidad_administrativa="Informática")
print(empleado1)  # Salida: Nombre: Juan Pérez, RUT: 12345678-9, Fecha de nacimiento: 1980-01-01, Unidad administrativa: Informática

empleado2 = Empleado.from_dict({"nombre": "María García", "rut": "98765432-1", "fecha_nacimiento": "1990-07-15", "unidad_administrativa": "Recursos Humanos"})
print(empleado2)  # Salida: Nombre: María García, RUT: 9876543
```

```python
# Ejercicio 3

from enum import Enum
import random

class Sexo(Enum):
    MASCULINO = 'M'
    FEMENINO = 'F'

class Persona:
    def __init__(self, dni, nombre_completo="", edad=0, sexo=Sexo.FEMENINO, peso=0, altura=0):
        self.nombre_completo = nombre_completo
        self.edad = edad
        self.dni = self._genera_dni
        self.sexo = sexo
        self.peso = peso
        self.altura = altura

    @property
    def nombre_completo(self):
        return self._nombre_completo

    @nombre_completo.setter
    def nombre_completo(self, nombre_completo):
        self._nombre_completo = nombre_completo

    @property
    def edad(self):
        return self._edad

    @edad.setter
    def edad(self, edad):
        if edad < 0:
            raise ValueError("La edad no puede ser negativa")
        self._edad = edad

    def calcular_imc(self):
        return self.peso / (self.altura ** 2)

    def valorar_imc(self):
        if self.calcular_imc() < 18:
            return -1
        elif self.calcular_imc() > 25:
            return 1

        return 0

    def mayor_edad(self):
        return self.edad >= 18

    def _genera_dni():
        return str(random.randint(10000000, 99999999))

    def __str__(self):
        return f"Nombre: {self.nombre_completo}, Edad: {self.edad}, DNI: {self.dni}, Sexo: {self.sexo}, Peso: {self.peso}, Altura: {self.altura}"

    def __repr__(self):
        return f"self.__class__.name (nombre_completo={self.nombre_completo}, edad={self.edad}, dni={self.dni}, sexo={self.sexo}, peso={self.peso}, altura={self.altura})"
```

```python
# Ejercicio 4

import random

class Empleado:
  """Clase que representa a un empleado de la empresa."""

  def __init__(self, nombre, rut, sueldo_base=1000000):
    """Inicializa un nuevo empleado.

    Args:
      nombre: El nombre del empleado.
      rut: El RUT del empleado.
      sueldo_base: El sueldo base del empleado.
    """
    self.nombre = nombre
    self.rut = rut
    self.sueldo_base = sueldo_base

  def calcular_sueldo(self):
    """Calcula el sueldo del empleado."""
    return self.sueldo_base

class Jefe(Empleado):
  """Clase que representa a un jefe de la empresa."""

  def __init__(self, nombre, rut, sueldo_base=1000000):
    """Inicializa un nuevo jefe.

    Args:
      nombre: El nombre del jefe.
      rut: El RUT del jefe.
      sueldo_base: El sueldo base del empleado.
      antiguedad: La antigüedad del jefe en años.
    """
    super().__init__(nombre, rut, sueldo_base)
    self.antiguedad = random.randint(0, 10)

  def calcular_sueldo(self):
    """Calcula el sueldo del jefe.

    Returns:
      El sueldo del jefe.
    """
    return self.antiguedad * 10 + self.sueldo_base

class Viajante(Empleado):
  """Clase que representa a un viajante de la empresa."""

  def __init__(self, nombre, rut, sueldo_base=1000000):
    """Inicializa un nuevo viajante."""
    super().__init__(nombre, rut, sueldo_base)
    self.numero_viajes = random.randint(0, 10)

  def calcular_sueldo(self):
    """Calcula el sueldo del viajante.

    Returns:
      El sueldo del viajante.
    """
    return self.sueldo_base + self.numero_viajes * 300

class Empresa:
  """Clase que representa una empresa."""
  def __init__(self, nombre):
    """Inicializa una nueva empresa.

    Args:
      nombre: El nombre de la empresa.
    """
    self.nombre = nombre
    self.empleados = []

  def agregar_empleado(self, empleado):
    """Agrega un empleado a la empresa.

    Args:
      empleado: El empleado a agregar.
    """
    self.empleados.append(empleado)

empresa = Empresa("UCM")

empresa.agregar_empleado(Jefe("Juan Pérez", "12345678-9"))
empresa.agregar_empleado(Jefe("María García", "98765432-1"))

for i in range(5):
  viajante = Empleado(f"Viajante {i+1}", f"1234567{i+1}-9")
  empresa.agregar_empleado(viajante)

for i in range(15):
  empleado = Empleado(f"Empleado {i+1}", f"1234567{i+1}-9")
  empresa.agregar_empleado(empleado)
```

```python
# Ejercicio 5

import datetime


class CuentaBancaria:
  """Clase que representa una cuenta bancaria."""

  def __init__(self, nombre_titular, fecha_apertura, numero_cuenta, saldo=0):
    """Inicializa una nueva cuenta bancaria.

    Args:
      nombre_titular: El nombre del titular de la cuenta.
      fecha_apertura: La fecha de apertura de la cuenta en formato yyyy-mm-dd.
      numero_cuenta: El número de cuenta.
      saldo: El saldo inicial de la cuenta.
    """
    self.nombre_titular = nombre_titular
    self.fecha_apertura = datetime.datetime.strptime(fecha_apertura, "%Y-%m-%d")
    self.numero_cuenta = numero_cuenta
    self.saldo = saldo

  def retirar_dinero(self, monto):
    """Retira dinero de la cuenta.

    Args:
      monto: La cantidad de dinero a retirar.

    Raises:
      ValueError: Si el monto a retirar es mayor que el saldo.
    """
    if monto > self.saldo:
      raise ValueError("Saldo insuficiente")
    self.saldo -= monto

  def ingresar_dinero(self, monto):
    """Ingresa dinero a la cuenta.

    Args:
      monto: La cantidad de dinero a ingresar.
    """
    self.saldo += monto

  def transferir_dinero(self, cuenta_destino, monto):
    """Transfiere dinero a otra cuenta.

    Args:
      cuenta_destino: La cuenta de destino a la que se transfiere el dinero.
      monto: La cantidad de dinero a transferir.

    Raises:
      ValueError: Si el monto a transferir es mayor que el saldo.
    """
    if monto > self.saldo:
      raise ValueError("Saldo insuficiente")
    self.saldo -= monto
    cuenta_destino.saldo += monto


class CuentaPlazoFijo(CuentaBancaria):
  """Clase que representa una cuenta a plazo fijo."""

  def __init__(self, nombre_titular, fecha_apertura, numero_cuenta, saldo, fecha_vencimiento):
    """Inicializa una nueva cuenta a plazo fijo.

    Args:
      nombre_titular: El nombre del titular de la cuenta.
      fecha_apertura: La fecha de apertura de la cuenta en formato yyyy-mm-dd.
      numero_cuenta: El número de cuenta.
      saldo: El saldo inicial de la cuenta.
      fecha_vencimiento: La fecha de vencimiento de la cuenta en formato yyyy-mm-dd.
    """
    super().__init__(nombre_titular, fecha_apertura, numero_cuenta, saldo)
    self.fecha_vencimiento = datetime.datetime.strptime(fecha_vencimiento, "%Y-%m-%d")

  def retirar_dinero(self, monto):
    """Retira dinero de la cuenta.

    Args:
      monto: La cantidad de dinero a retirar.

    Raises:
      ValueError: Si el monto a retirar es mayor que el saldo.
    """
    if datetime.datetime.today() < self.fecha_vencimiento:
      monto_penalizacion = monto * 0.05
      monto_a_retirar = monto + monto_penalizacion
      if monto_a_retirar > self.saldo:
        raise ValueError("Saldo insuficiente")
      self.saldo -= monto_a_retirar
    else:
      super().retirar_dinero(monto)


class CuentaVip(CuentaBancaria):
  """Clase que representa una cuenta Vip."""

  def __init__(self, nombre_titular, fecha_apertura, numero_cuenta, saldo, saldo_negativo_maximo):
    """Inicializa una nueva cuenta Vip.

    Args:
      nombre_titular: El nombre del titular de la cuenta.
      fecha_apertura: La fecha de apertura de la cuenta en formato yyyy-mm-dd.
      numero_cuenta: El número de cuenta.
      saldo: El saldo inicial de la cuenta.
      saldo_negativo_maximo: El saldo negativo máximo permitido.
    """
    super().__init__(nombre_titular, fecha_apertura, numero_cuenta, saldo)
    self.saldo_negativo_maximo = saldo_negativo_maximo

  def retirar_dinero(self, monto):
    """Retira dinero de la cuenta.

    Args:
      monto: La cantidad de dinero a retirar.
    """
    nuevo_saldo = self.saldo - monto
    if nuevo_saldo < -self.saldo_negativo_maximo:
      raise ValueError("Saldo negativo excedido")
    self.saldo = nuevo_saldo

class Banco:
  """Clase que representa un banco."""
  def __init__(self, nombre):
    """Inicializa un nuevo banco.

    Args:
      nombre: El nombre del banco.
    """
    self.nombre = nombre
    self.cuentas = []

  def agregar_cuenta(self, cuenta):
    """Agrega una cuenta al banco.

    Args:
      cuenta: La cuenta a agregar.
    """
    self.cuentas.append(cuenta)

banco = Banco("Banco XYZ")

for i in range(9):
  cuenta = CuentaBancaria(f"Titular {i+1}", f"2023-0{i+1}-01", f"1234567{i+1}")
  banco.agregar_cuenta(cuenta)

for i in range(5):
  cuenta = CuentaPlazoFijo(f"Titular {i+1}", f"2023-0{i+1}-01", f"1234567{i+1}",
                            saldo=1000000, fecha_vencimiento="2024-01-01")
  banco.agregar_cuenta(cuenta)

for i in range(5):
  cuenta = CuentaVip(f"Titular {i+1}", f"2023-0{i+1}-01", f"1234567{i+1}",
                            saldo=1000000, saldo_negativo_maximo=i*10000)
  banco.agregar_cuenta(cuenta)
```
